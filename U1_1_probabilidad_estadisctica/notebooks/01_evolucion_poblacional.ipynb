# Actividad 3 — Estadística descriptiva 

```python
# Selección de un año histórico
año_seleccionado = 2020
```

```python
# Filtrar datos del año y eliminar valores faltantes
df_año = df[df["Year"] == año_seleccionado].dropna(
    subset=["Annual change in population"]
)
```

```python

variable = df_año["Annual change in population"]
```

```python
# Estadísticos descriptivos
media = variable.mean()
mediana = variable.median()
varianza = variable.var()
desviacion_estandar = variable.std()
minimo = variable.min()
maximo = variable.max()
```

```python
print(f"Estadísticos para el año {año_seleccionado}")
print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion_estandar)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
# Entidades con el valor mínimo y máximo
entidad_minima = df_año.loc[
    variable.idxmin(), ["Entity", "Annual change in population"]
]

entidad_maxima = df_año.loc[
    variable.idxmax(), ["Entity", "Annual change in population"]
]

print("Entidad con valor mínimo:")
print(entidad_minima)

print("\nEntidad con valor máximo:")
print(entidad_maxima)
```

```python
#Sacar estadisticos de:
# -estadisticas Pais 1er letra del apellido(Z) 1983-2021
# -df minimo-máximo

paises_con_z = df[df["Entity"].str.startswith("Z", na=False)]["Entity"].unique()
print("Países que inician con Z:")
print(paises_con_z)
```

```python
#elegimos el pais
pais_z = "Zambia" 
```

```python
# Filtramos por país y por el rango de años
df_z = df[(df["Entity"] == pais_z) & (df["Year"] >= 1983) & (df["Year"] <= 2021)]
```

```python
# Sacamos las estadísticas de la variable que nos interesa
estadisticas_z = df_z["Annual change in population"].describe()
print(f"--- Estadísticas para {pais_z} (1983-2021) ---")
print(estadisticas_z)
```

```python
variable_total = df["Annual change in population"].dropna()

fila_minima = df.loc[variable_total.idxmin(), ["Entity", "Year", "Annual change in population"]]
fila_maxima = df.loc[variable_total.idxmax(), ["Entity", "Year", "Annual change in population"]]

```

```python
print("Entidad con valor mínimo:")
print(fila_minima)

print("\nEntidad con valor máximo:")
print(fila_maxima)
```
