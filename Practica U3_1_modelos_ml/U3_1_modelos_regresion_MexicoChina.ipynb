## 5. Preparar los nombres de las columnas

Al explorar el dataset notamos dos cosas:

1. Los nombres de las columnas son largos y difíciles de escribir.
2. Los **datos históricos** (`..._variant_estimates`) y las **proyecciones de la ONU** (`..._variant_medium__projected`) vienen en **columnas separadas**. Cuando una fila es histórica, la columna de proyección está vacía (`NaN`), y viceversa.

Vamos a renombrar las columnas y a crear dos columnas unificadas, `births` y `deaths`, que contengan el dato histórico cuando existe y la proyección de OWID cuando no.

````python
df = df.rename(columns={
    "entity": "Entity",
    "code": "Code",
    "year": "Year",
    "births__sex_all__age_all__variant_estimates": "births_estimates",
    "births__sex_all__age_all__variant_medium__projected": "births_projected",
    "deaths__sex_all__age_all__variant_estimates": "deaths_estimates",
    "deaths__sex_all__age_all__variant_medium__projected": "deaths_projected",
})


df.head()
````

````python
df_inicial = df.copy()
df_inicial.head(3)
````

````python
# Columna unificada: dato histórico si existe; si no, la proyección de OWID
df["births"] = df["births_estimates"].fillna(df["births_projected"])
df["deaths"] = df["deaths_estimates"].fillna(df["deaths_projected"])
df.head()
````

# PARTE 5. Separar datos históricos y proyectados

## 10. Identificar el punto de separación

El dataset contiene **datos históricos** y **proyecciones**. Antes de separarlos, verifiquemos dónde termina uno y empieza el otro:

````python
ultimo_historico = df_mexico.loc[df_mexico["births_estimates"].notna(), "Year"].max()
primera_proyeccion = df_mexico.loc[df_mexico["births_projected"].notna(), "Year"].min()

print("Último año con dato histórico:", ultimo_historico)
print("Primer año con proyección de OWID:", primera_proyeccion)
````

Para esta práctica utilizaremos como punto de separación el **año 2023**. Creamos dos `DataFrames`:

````python
df_historico = df_mexico[
    df_mexico["Year"] <= 2023
]

df_proyectado = df_mexico[
    df_mexico["Year"] > 2023
]

print(df_historico.head(3))
print(df_proyectado.head(3))
````

````python
print("Histórico  ->", df_historico.shape, "| años:", df_historico["Year"].min(), "a", df_historico["Year"].max())
print("Proyectado ->", df_proyectado.shape, "| años:", df_proyectado["Year"].min(), "a", df_proyectado["Year"].max())
````

# PARTE 6. Preparar las variables para Machine Learning

## 11. Definir X e y

Para nuestro primer modelo utilizaremos:

* **X:** año → la **variable predictora**
* **y:** número de nacimientos → la **variable objetivo**

````python
X = df_historico[["Year"]]
y_nacimientos = df_historico["births"]
````

````python
X.head()
````

````python
y_nacimientos.head()
````

## 12. Convertir los datos a NumPy

Obtenemos los valores como arreglos de NumPy:

````python
X_numpy = X.to_numpy()
y_numpy = y_nacimientos.to_numpy()

print("X_numpy.shape:", X_numpy.shape)
print("y_numpy.shape:", y_numpy.shape)
````

````python
type(X_numpy), type(y_numpy)
````

### Observación

Scikit-learn espera normalmente que las variables predictoras tengan la forma:

```text
(número de observaciones, número de variables)
```
Es decir, que scikit-learn interpreta X como una tabla de datos.

Por eso `X` tiene **dos dimensiones**, mientras que `y` tiene una sola.

En nuestro ejemplo tenemos:

74 observaciones (registros).

1 variable predictora: año (estas son las caracteristicas o atributos usadas para la predicción).

1 variable objetivo: nacimientos (valor que deseamos predecir).

# PARTE 7. Modelo 1 – Regresión lineal

## 13. Crear el modelo

````python
modelo_lineal = LinearRegression()
````

## 14. Entrenar el modelo

Utilizamos los datos históricos para entrenar el modelo.

````python
modelo_lineal.fit(X, y_nacimientos)
````

Una regresión lineal aprende una recta `y = a·x + b`. Veamos qué valores aprendió:

````python
print("Pendiente (a):", modelo_lineal.coef_[0])
print("Intercepto (b):", modelo_lineal.intercept_)
````

## 15. Realizar predicciones

Utilizamos el modelo para estimar los nacimientos correspondientes a los **años históricos**.

````python
predicciones_lineal = modelo_lineal.predict(X)

predicciones_lineal[:10]
````

## 16. Realizar proyecciones hasta 2100

Creamos los años para los que queremos generar **nuestras propias proyecciones**.

````python
anios_futuros = np.arange(2024, 2101).reshape(-1, 1)

print("Forma:", anios_futuros.shape)
anios_futuros[:10]
````

> **Detalle técnico:** entrenamos el modelo con `X`, que es un `DataFrame` con una columna llamada `Year`. Scikit-learn recuerda ese nombre, así que para predecir también debemos entregarle un `DataFrame` con una columna `Year`. Si le pasáramos un arreglo de NumPy "a secas" mostraría una advertencia (*UserWarning: X does not have valid feature names*). Por eso convertimos los años futuros a `DataFrame`.

````python
X_futuro = pd.DataFrame(anios_futuros, columns=["Year"])

X_futuro.head()
````

````python
predicciones_lineal_futuras = modelo_lineal.predict(X_futuro)
````

````python
print(predicciones_lineal_futuras[:10])
````

# PARTE 8. Visualizar la regresión lineal

## 17. Comparar datos históricos y predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

# Ajuste del modelo sobre los años históricos
plt.plot(
    df_historico["Year"],
    predicciones_lineal,
    color="tab:orange",
    label="Regresión lineal (ajuste histórico)"
)

# Proyección hacia el futuro
plt.plot(
    anios_futuros.flatten(),
    predicciones_lineal_futuras,
    color="tab:orange",
    linestyle="--",
    label="Regresión lineal (proyección)"
)

plt.title("Proyección de nacimientos – Regresión lineal")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Analiza cómo continúa la tendencia histórica cuando utilizamos una **función lineal**. ¿Qué pasa con los valores hacia el año 2100? La función lineal proyecta erróneamente un crecimiento constante hasta el 2100, separándose por completo de la tendencia real de los datos.

# PARTE 9. Modelo 2 – Regresión polinomial

## 18. Crear variables polinomiales

La regresión lineal solamente puede representar una relación **lineal** entre las variables. Ahora construiremos variables polinomiales para permitir **curvas**. Utilizaremos un polinomio de **grado 2**:

$$y = a \cdot x^2 + b \cdot x + c$$

````python
poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)
````

Transformamos `X`:

````python
X_poly = poly.fit_transform(X)

X_poly[:5]
````

Observa que ahora cada año tiene **dos columnas**: el año (`x`) y el año al cuadrado (`x²`).

## 19. Crear y entrenar el modelo polinomial

Seguimos utilizando `LinearRegression`; lo único que cambió son las **variables de entrada**.

````python
modelo_polinomial = LinearRegression()

modelo_polinomial.fit(
    X_poly,
    y_nacimientos
)
````

## 20. Generar predicciones históricas

````python
predicciones_polinomiales = modelo_polinomial.predict(
    X_poly
)
````

## 21. Generar proyecciones hasta 2100

Transformamos los años futuros utilizando **el mismo objeto `poly`** con el que transformamos los datos de entrenamiento.

````python
anios_futuros_poly = poly.transform(
    X_futuro
)

predicciones_polinomiales_futuras = modelo_polinomial.predict(
    anios_futuros_poly
)
````

# PARTE 10. Visualizar la regresión polinomial

## 22. Graficar las predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

plt.plot(
    df_historico["Year"],
    predicciones_polinomiales,
    color="tab:green",
    label="Regresión polinomial (ajuste histórico)"
)

plt.plot(
    anios_futuros.flatten(),
    predicciones_polinomiales_futuras,
    color="tab:green",
    linestyle="--",
    label="Regresión polinomial (proyección)"
)

plt.title("Proyección de nacimientos – Regresión polinomial")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Compara el comportamiento de esta gráfica con el modelo de regresión lineal. ¿Se ajusta mejor a los datos históricos? Sí. Al permitir curvatura, el modelo logra seguir con bastante precisión la etapa histórica: captura muy bien cómo los nacimientos subieron durante décadas y llegaron a su punto máximo, a diferencia de la línea recta de la regresión lineal. ¿Qué ocurre con la proyección hacia el año 2100? Aquí es donde el modelo falla drásticamente. Las funciones matemáticas polinomiales tienden a irse a los extremos cuando se "extrapolan". ¿Los valores que predice tienen sentido para un número de nacimientos? No, no tienen ningún sentido. Como la curva cae en picada y no tiene un límite inferior, el modelo polinomial eventualmente terminará prediciendo valores negativos para los nacimientos hacia o antes del 2100.

# PARTE 11. Modelo 3 – Random Forest

## 23. Crear el modelo

Un **Random Forest** está formado por muchos árboles de decisión. Utilizaremos `RandomForestRegressor`.

````python
modelo_rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
````

## 24. Entrenar el modelo

````python
modelo_rf.fit(
    X,
    y_nacimientos
)
````

## 25. Generar predicciones históricas

````python
predicciones_rf = modelo_rf.predict(X)
````

## 26. Generar proyecciones futuras

````python
predicciones_rf_futuras = modelo_rf.predict(
    X_futuro
)
````

# PARTE 12. Visualizar Random Forest

## 27. Graficar las predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

plt.plot(
    df_historico["Year"],
    predicciones_rf,
    color="tab:red",
    label="Random Forest (ajuste histórico)"
)

plt.plot(
    anios_futuros.flatten(),
    predicciones_rf_futuras,
    color="tab:red",
    linestyle="--",
    label="Random Forest (proyección)"
)

plt.title("Proyección de nacimientos – Random Forest")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Fíjate en dos cosas: (1) qué tan bien sigue el modelo los datos históricos y (2) qué sucede con la línea después de 2023. ¿Continúa la tendencia? ¿Por qué crees que sucede esto? Investiga. 

¿Qué tan bien sigue el modelo los datos históricos? Los sigue perfectamente (a veces demasiado bien, lo que se conoce como sobreajuste o overfitting). Como el modelo Random Forest crea múltiples árboles de decisión que aprenden a partir de los datos exactos, la línea de predicción del modelo se superpondrá casi como un calco exacto de los datos históricos. Captura cada pequeño salto y caída que hubo en el pasado.

¿Qué sucede con la línea después de 2023? ¿Continúa la tendencia? La línea se vuelve completamente plana (horizontal). No continúa la tendencia ni hacia arriba ni hacia abajo, sin importar que estemos intentando predecir hasta el año 2100.

### 1.
¿Qué diferencia observaste entre la regresión lineal, la regresión polinomial y Random Forest?

*Respuesta: La principal diferencia es cómo se adaptan a la forma de los datos y cómo predicen hacia el futuro. La regresión lineal asume un crecimiento/decrecimiento constante creando una línea completamente recta que no sigue las curvas reales. La regresión polinomial se ajusta muy bien a las curvas históricas (crecimiento y caída), pero al predecir a futuro se descontrola y cae abruptamente hacia valores irreales. Random Forest copia a la perfección cada variación del pasado, pero se vuelve una línea horizontal completamente plana hacia el futuro porque no sabe proyectar tendencias nuevas.* 

### 2.
¿Por qué los tres modelos generan predicciones diferentes si utilizaron los mismos datos históricos?

*Respuesta: Porque cada modelo utiliza algoritmos y matemáticas diferentes que asumen comportamientos distintos sobre cómo funcionan los datos. El lineal busca la línea recta que pase por el medio de los puntos; el polinomial busca una ecuación curva que conecte la mayor cantidad de puntos posibles; y Random Forest no usa ecuaciones, sino que agrupa los datos mediante árboles de decisiones lógicas basadas únicamente en lo que ya ocurrió.*

### 3.
¿Qué modelo produjo una tendencia más cercana visualmente a las proyecciones de OWID?

Sin embargo, a muy largo plazo se separa drásticamente al caer en picada.*

*Respuesta: No necesariamente. El parecido visual puede ser coincidencia o producto de un "sobreajuste". Para determinar si un modelo es bueno, no basta con verlo; se requiere evaluarlo numéricamente con métricas de error y asegurarse de que sus predicciones no violen la lógica del mundo real.*

### 5.
¿Qué comportamiento observaste en las predicciones de Random Forest para los años posteriores a 2023?

*Respuesta: La predicción se convierte en una línea recta horizontal. El modelo repite exactamente el mismo valor proyectado para todos los años futuros porque es incapaz de extrapolar datos que están fuera del rango de tiempo con el que fue entrenado*

# EJERCICIO 1 — China: Nacimientos y Defunciones (todos los años)

Extraemos los datos de **China** del dataset y aplicamos los mismos tres modelos de regresión:

1. Regresión Lineal (RL)
2. Regresión Polinomial (RP)
3. Random Forest Regressor (RF)

Se entrena con datos históricos (`_estimates`) y se predicen los años completos (1950–2100).

````python

df_china = df[df["Entity"] == "China"].copy()

print(f"Rango de años: {df_china['Year'].min()} - {df_china['Year'].max()}")
print(f"Número de filas: {df_china.shape[0]}")
df_china.head()
````

````python

df_china_hist = df_china[df_china["births_estimates"].notna()].copy()
df_china_proy = df_china.copy()

X_china_hist = df_china_hist[["Year"]].values
y_china_births_hist = df_china_hist["births"].values
y_china_deaths_hist = df_china_hist["deaths"].values

X_china_todos = df_china_proy[["Year"]].values

print(f"Años históricos: {df_china_hist['Year'].min()} - {df_china_hist['Year'].max()} ({len(df_china_hist)} filas)")
````

````python


# — Regresión Lineal —
rl_china_births = LinearRegression()
rl_china_births.fit(X_china_hist, y_china_births_hist)

rl_china_deaths = LinearRegression()
rl_china_deaths.fit(X_china_hist, y_china_deaths_hist)


poly_china = PolynomialFeatures(degree=3)
X_china_hist_poly = poly_china.fit_transform(X_china_hist)
X_china_todos_poly = poly_china.transform(X_china_todos)

rp_china_births = LinearRegression()
rp_china_births.fit(X_china_hist_poly, y_china_births_hist)

rp_china_deaths = LinearRegression()
rp_china_deaths.fit(X_china_hist_poly, y_china_deaths_hist)


rf_china_births = RandomForestRegressor(n_estimators=100, random_state=42)
rf_china_births.fit(X_china_hist, y_china_births_hist)

rf_china_deaths = RandomForestRegressor(n_estimators=100, random_state=42)
rf_china_deaths.fit(X_china_hist, y_china_deaths_hist)

print("Modelos entrenados correctamente para China.")
````

````python

pred_china_rl_births = rl_china_births.predict(X_china_todos)
pred_china_rp_births = rp_china_births.predict(X_china_todos_poly)
pred_china_rf_births = rf_china_births.predict(X_china_todos)

pred_china_rl_deaths = rl_china_deaths.predict(X_china_todos)
pred_china_rp_deaths = rp_china_deaths.predict(X_china_todos_poly)
pred_china_rf_deaths = rf_china_deaths.predict(X_china_todos)

anios_china = df_china_proy["Year"].values
print("Predicciones generadas.")
````

````python

fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(df_china_proy["Year"], df_china_proy["births"],
        color="black", linewidth=2, label="OWID (histórico + proyección)", zorder=5)
ax.scatter(df_china_hist["Year"], df_china_hist["births"],
           color="black", s=15, zorder=6, label="Datos históricos (entrenamiento)")

ax.plot(anios_china, pred_china_rl_births,
        color="royalblue", linestyle="--", linewidth=1.8, label="Regresión Lineal (RL)")
ax.plot(anios_china, pred_china_rp_births,
        color="darkorange", linestyle="-.", linewidth=1.8, label="Regresión Polinomial (RP)")
ax.plot(anios_china, pred_china_rf_births,
        color="seagreen", linestyle=":", linewidth=2.0, label="Random Forest (RF)")

ax.axvline(x=2023, color="gray", linestyle="--", alpha=0.6, label="Inicio proyección (2023)")

ax.set_title("China — Nacimientos: comparación de modelos (RL, RP, RF)", fontsize=14, fontweight="bold")
ax.set_xlabel("Año")
ax.set_ylabel("Nacimientos")
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x:,.0f}"))
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

````python

fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(df_china_proy["Year"], df_china_proy["deaths"],
        color="black", linewidth=2, label="OWID (histórico + proyección)", zorder=5)
ax.scatter(df_china_hist["Year"], df_china_hist["deaths"],
           color="black", s=15, zorder=6, label="Datos históricos (entrenamiento)")

ax.plot(anios_china, pred_china_rl_deaths,
        color="royalblue", linestyle="--", linewidth=1.8, label="Regresión Lineal (RL)")
ax.plot(anios_china, pred_china_rp_deaths,
        color="darkorange", linestyle="-.", linewidth=1.8, label="Regresión Polinomial (RP)")
ax.plot(anios_china, pred_china_rf_deaths,
        color="seagreen", linestyle=":", linewidth=2.0, label="Random Forest (RF)")

ax.axvline(x=2023, color="gray", linestyle="--", alpha=0.6, label="Inicio proyección (2023)")

ax.set_title("China — Defunciones: comparación de modelos (RL, RP, RF)", fontsize=14, fontweight="bold")
ax.set_xlabel("Año")
ax.set_ylabel("Defunciones")
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x:,.0f}"))
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````
